```python
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
tokenizer = AutoTokenizer.from_pretrained('../models/DeepSeek-R1-Distill-Qwen-7B')
```

```python
messages = [
     [
        {
            "role": "system",
            "content": "You are a friendly chatbot who always responds in the style of a pirate",
        },
            {"role": "user", "content": "How many helicopters can a human eat in one sitting?"},
     ],
     [
        {
            "role": "system",
            "content": "You are a friendly chatbot who always responds in the style of a pirate",
        },
            {"role": "user", "content": "你是谁?"},
     ],
]
```

```python
formatted_text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
input = tokenizer(formatted_text, padding=True, return_tensors='pt').to('cuda')
input_ids = input['input_ids']
attn_mask = input['attention_mask']
print(input_ids)
print(attn_mask)
print(tokenizer.convert_ids_to_tokens(input_ids[1]))
print(tokenizer.convert_ids_to_tokens(input_ids[1]))
```

```python
model = AutoModelForCausalLM.from_pretrained('../models/DeepSeek-R1-Distill-Qwen-7B').half().to('cuda')
```

```python
out = model.generate(input_ids, attention_mask=attn_mask, max_new_tokens=1024, do_sample=False)
# print(out)
# print(tokenizer.convert_ids_to_tokens(out[1]))
print(tokenizer.decode(out[1][input_ids.shape[1]:]))
```
